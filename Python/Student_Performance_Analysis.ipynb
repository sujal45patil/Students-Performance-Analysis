```python

import os
import matplotlib.pyplot as plt
import seaborn as sns

from reportlab.platypus import (
    SimpleDocTemplate, Paragraph, Spacer,
    Table, TableStyle, Image, PageBreak
)
from reportlab.lib import colors
from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle
from reportlab.lib.enums import TA_CENTER
from reportlab.lib.units import inch
from reportlab.lib.pagesizes import A4

# ---------------------------------
# 1. Create output folder
# ---------------------------------

output_folder = "Student_Performance_Report"
os.makedirs(output_folder, exist_ok=True)

# ---------------------------------
# 2. Generate charts
# ---------------------------------

sns.set_style("whitegrid")

# Chart 1: Exam Score Distribution
plt.figure(figsize=(7, 4))
sns.histplot(data=cleaned_df, x="Exam_Score", bins=15, kde=True)
plt.title("Exam Score Distribution")
plt.tight_layout()
plt.savefig(f"{output_folder}/exam_score.png")
plt.close()

# Chart 2: Performance Distribution
plt.figure(figsize=(7, 4))
sns.countplot(data=cleaned_df, x="Performance")
plt.title("Student Performance Distribution")
plt.tight_layout()
plt.savefig(f"{output_folder}/performance.png")
plt.close()

# Chart 3: Pass Status
plt.figure(figsize=(6, 4))
sns.countplot(data=cleaned_df, x="Pass_Status")
plt.title("Pass Status Distribution")
plt.tight_layout()
plt.savefig(f"{output_folder}/pass_status.png")
plt.close()

# Chart 4: Study Hours vs Exam Score
plt.figure(figsize=(7, 4))
sns.scatterplot(
    data=cleaned_df,
    x="Hours_Studied",
    y="Exam_Score"
)
plt.title("Study Hours vs Exam Score")
plt.tight_layout()
plt.savefig(f"{output_folder}/study_hours.png")
plt.close()

# Chart 5: Attendance vs Exam Score
plt.figure(figsize=(7, 4))
sns.scatterplot(
    data=cleaned_df,
    x="Attendance",
    y="Exam_Score"
)
plt.title("Attendance vs Exam Score")
plt.tight_layout()
plt.savefig(f"{output_folder}/attendance.png")
plt.close()

# ---------------------------------
# 3. Create PDF document
# ---------------------------------

pdf_path = "Student_Performance_Report.pdf"

doc = SimpleDocTemplate(
    pdf_path,
    pagesize=A4,
    rightMargin=40,
    leftMargin=40,
    topMargin=40,
    bottomMargin=40
)

styles = getSampleStyleSheet()

title_style = ParagraphStyle(
    "ReportTitle",
    parent=styles["Title"],
    alignment=TA_CENTER,
    fontSize=20,
    spaceAfter=20
)

heading_style = styles["Heading2"]
normal_style = styles["BodyText"]

story = []

# ---------------------------------
# 4. Report Title
# ---------------------------------

story.append(Paragraph(
    "STUDENT PERFORMANCE ANALYSIS REPORT",
    title_style
))

story.append(Paragraph(
    "Automated Data Analysis and Business Insights",
    ParagraphStyle(
        "Subtitle",
        parent=normal_style,
        alignment=TA_CENTER
    )
))

story.append(Spacer(1, 20))

# ---------------------------------
# 5. Dataset Overview
# ---------------------------------

story.append(Paragraph("1. Dataset Overview", heading_style))

overview_data = [
    ["Metric", "Value"],
    ["Total Students", str(cleaned_df.shape[0])],
    ["Total Columns", str(cleaned_df.shape[1])],
    ["Missing Values", str(cleaned_df.isnull().sum().sum())],
    ["Duplicate Records", str(cleaned_df.duplicated().sum())]
]

overview_table = Table(overview_data, colWidths=[220, 180])

overview_table.setStyle(TableStyle([
    ("BACKGROUND", (0, 0), (-1, 0), colors.lightblue),
    ("GRID", (0, 0), (-1, -1), 0.5, colors.grey),
    ("PADDING", (0, 0), (-1, -1), 8),
    ("FONTNAME", (0, 0), (-1, 0), "Helvetica-Bold")
]))

story.append(overview_table)
story.append(Spacer(1, 20))

# ---------------------------------
# 6. Statistical Summary
# ---------------------------------

story.append(Paragraph("2. Statistical Summary", heading_style))

numeric_summary = cleaned_df.describe().round(2).reset_index()

stat_data = [numeric_summary.columns.tolist()]

for row in numeric_summary.values.tolist():
    stat_data.append([str(value) for value in row])

stat_table = Table(stat_data, repeatRows=1)

stat_table.setStyle(TableStyle([
    ("BACKGROUND", (0, 0), (-1, 0), colors.lightblue),
    ("GRID", (0, 0), (-1, -1), 0.4, colors.grey),
    ("FONTSIZE", (0, 0), (-1, -1), 7),
    ("PADDING", (0, 0), (-1, -1), 4)
]))

story.append(stat_table)
story.append(Spacer(1, 20))

# ---------------------------------
# 7. Categorical Summary
# ---------------------------------

story.append(PageBreak())
story.append(Paragraph("3. Categorical Summary", heading_style))

categorical_summary = cleaned_df.describe(
    include="object"
).T.reset_index()

categorical_summary = categorical_summary.fillna("")

cat_data = [["Column", "Unique Values", "Most Frequent", "Frequency"]]

for _, row in categorical_summary.iterrows():
    cat_data.append([
        str(row["index"]),
        str(row["unique"]),
        str(row["top"]),
        str(row["freq"])
    ])

cat_table = Table(
    cat_data,
    colWidths=[145, 75, 130, 65],
    repeatRows=1
)

cat_table.setStyle(TableStyle([
    ("BACKGROUND", (0, 0), (-1, 0), colors.lightblue),
    ("GRID", (0, 0), (-1, -1), 0.4, colors.grey),
    ("FONTSIZE", (0, 0), (-1, -1), 7),
    ("PADDING", (0, 0), (-1, -1), 5)
]))

story.append(cat_table)

# ---------------------------------
# 8. Charts
# ---------------------------------

story.append(PageBreak())
story.append(Paragraph("4. Data Visualizations", heading_style))

chart_files = [
    ("Exam Score Distribution", "exam_score.png"),
    ("Student Performance Distribution", "performance.png"),
    ("Pass Status Distribution", "pass_status.png"),
    ("Study Hours vs Exam Score", "study_hours.png"),
    ("Attendance vs Exam Score", "attendance.png")
]

for chart_title, chart_file in chart_files:
    story.append(Paragraph(chart_title, styles["Heading3"]))

    chart_path = os.path.join(output_folder, chart_file)

    story.append(Image(
        chart_path,
        width=6 * inch,
        height=3.4 * inch
    ))

    story.append(Spacer(1, 12))

# ---------------------------------
# 9. Business Insights
# ---------------------------------

story.append(PageBreak())
story.append(Paragraph("5. Business Insights", heading_style))

average_score = cleaned_df["Exam_Score"].mean()
average_attendance = cleaned_df["Attendance"].mean()
average_study = cleaned_df["Hours_Studied"].mean()

pass_count = (
    cleaned_df["Pass_Status"].astype(str).str.strip().str.lower()
    == "pass"
).sum()

fail_count = len(cleaned_df) - pass_count

insights = [
    f"The average exam score of students is {average_score:.2f}.",
    f"The average attendance is {average_attendance:.2f}%.",
    f"Students study an average of {average_study:.2f} hours.",
    f"A total of {pass_count} students are marked as Pass.",
    f"A total of {fail_count} students are marked as not Pass."
]

for insight in insights:
    story.append(Paragraph("• " + insight, normal_style))
    story.append(Spacer(1, 10))

# ---------------------------------
# 10. Conclusion
# ---------------------------------

story.append(Spacer(1, 15))
story.append(Paragraph("6. Conclusion", heading_style))

conclusion = """
This report provides an overview of student performance using
statistical analysis, categorical analysis and data visualizations.
The findings can help educational institutions understand student
outcomes and identify areas where academic support may be useful.
"""

story.append(Paragraph(conclusion, normal_style))

# ---------------------------------
# 11. Generate PDF
# ---------------------------------

doc.build(story)

print("PDF Report Generated Successfully!")
print("File Name:", pdf_path)
print("Location:", os.path.abspath(pdf_path))
```
